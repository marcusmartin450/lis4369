## 15. Rename Multiple Columns

```python
column_mapping = {
    'Postcode': 'Zip',
    'Nursing': 'Nursing_mins',
    'Laboratory': 'Laboratory_mins',
    'Consultation': 'Consultation_mins',
    'Pharmacy': 'Pharmacy_mins'
}
df_patients.rename(columns=column_mapping, inplace=True)
df_patients.head()
```
