This notebook demonstrates how to clean, explore, and analyze patient data using **pandas** and **matplotlib** in Python.  
We will check for missing values, rename columns, save cleaned data, and visualize distributions.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load Dataset
We begin by loading the patient dataset from a CSV file.

```
df_patients = pd.read_csv("patients.csv")
```

## 2. Inspect Dataset
Check dataset information and preview some rows.

```
df_patients.info()
df_patients.head()
df_patients.tail()
```

## 3. Check Missing Values
Use `.isna()` to detect missing values.  
- `True` means missing  
- `False` means not missing

```
df_patients.isna()
```

### Count of Missing Values per Column

```
df_patients.isna().sum()
```

## 4. Drop Missing Values
If missing values exist, we can drop them with `.dropna()`.  
Here, the dataset has no missing values, but we demonstrate the command.

```
df_patients = df_patients.dropna()
df_patients.head()
```

## 5. Display Records
Show sample records for inspection.

```
# First 5 records
df_patients.head()

# Last 5 records
df_patients.tail()

# First 10 records
df_patients.head(10)
```

## 6. Rename Columns
Some column names are not intuitive. We’ll rename them for clarity.

```
df_patients.rename(columns={'Registration': 'Registration_mins'})
```

### Make Rename Permanent
Use `inplace=True` to apply changes permanently.

```
df_patients.rename(columns={'Registration': 'Registration_mins'}, inplace=True)
```

### Rename Multiple Columns
We can rename multiple columns using a dictionary.

```
column_mapping = {
    'Postcode': 'Zip',
    'Nursing': 'Nursing_mins',
    'Laboratory': 'Laboratory_mins',
    'Consultation': 'Consultation_mins',
    'Pharmacy': 'Pharmacy_mins'
}
df_patients.rename(columns=column_mapping, inplace=True)
df_patients.head()
```

## 7. Save Cleaned Data
Save the cleaned dataset to a new CSV file for later use.

```
df_patients.to_csv("patients_clean.csv")
```

### Read Cleaned Data
Load the cleaned file to confirm changes.

```
pd.read_csv("patients_clean.csv").head()
```

## 8. Quick Stats
Use `.describe()` to view summary statistics for numeric columns.

```
df_patients.describe()
```

## 9. Visualization
Plot the distribution of patients with and without insurance.

```
patient_insurance = df_patients['Has_Insurance'].value_counts()
patient_insurance.plot(kind='bar')
plt.show()
```
