```python
print("\nProgram Requirements:\n")
print("1. Must use float data type for user input (except, 'Party Number').")
print("2. Must round calculations to two decimal places.")
print("3. Must format currency with dollar sign, and two decimal places.\n")
```

```python
meal_cost = float(input("Cost of meal: "))
tax_percent = float(input("Tax percent: "))
tip_percent = float(input("Tip percent: "))
people_num = int(input("Party number: "))
```

```python
tax_amount = round(meal_cost * (tax_percent / 100), 2)      # tax
tip_amount = round(meal_cost * (tip_percent / 100), 2)      # tip
total_amount = round(meal_cost + tax_amount + tip_amount, 2)
split = round(total_amount / people_num, 2)
```

```python
print("\nProgram Output:")
print("Meal cost: ${:,.2f}".format(meal_cost))
print("Tax: ${:,.2f}".format(tax_amount))
print("Tip amount: ${:,.2f}".format(tip_amount))
print("Total: ${:,.2f}".format(total_amount))
print("Split = " + str(people_num) + " -> ${:,.2f}".format(split))
```
